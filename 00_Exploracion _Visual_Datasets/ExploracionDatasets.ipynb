## INDENTIFICACIÓN DE NULOS

```python
# Nulos
def reporte_nulos(df, nombre):
    nulos = df.isnull().sum()
    porcentaje = (nulos / len(df) * 100).round(2)
    reporte = pd.DataFrame({"nulos": nulos, "porcentaje": porcentaje})
    reporte = reporte[reporte["nulos"] > 0].sort_values("nulos", ascending=False)
    print(f"\n NULOS EN {nombre}:")
    print(reporte if not reporte.empty else "Sin nulos detectados")

reporte_nulos(la_libertad, "ATENCIONES EN LA LIBERTAD SIS")
reporte_nulos(renipress, "RENIPRESS")

# Duplicados
print(f"\nDuplicados totales en Atenciones: {atenciones.duplicated().sum()}")
print(f"Duplicados totales en RENIPRESS: {renipress.duplicated().sum()}")
print(f"Duplicados por CODIGO_IPRESS en RENIPRESS: {renipress.duplicated(subset=['COD_IPRESS']).sum()}")
```

### Validación de integridad referencial: cruce COD_IPRESS (Atenciones) y RENIPRESS

```python
la_libertad["COD_IPRESS"] = la_libertad["COD_IPRESS"].astype(str).str.strip()
renipress["COD_IPRESS"] = renipress["COD_IPRESS"].astype(str).str.strip()

print("Longitud de códigos (para detectar ceros a la izquierda faltantes)\n")
print("Atenciones COD_IPRESS:")
print(la_libertad["COD_IPRESS"].str.len().value_counts())
print("\nRENIPRESS COD_IPRESS:")
print(renipress["COD_IPRESS"].str.len().value_counts())

codigos_atenciones = set(la_libertad["COD_IPRESS"])
codigos_renipress = set(renipress["COD_IPRESS"])

sin_match = codigos_atenciones - codigos_renipress
print(f"\nCódigos únicos en Atenciones: {len(codigos_atenciones)}")
print(f"Códigos únicos en RENIPRESS: {len(codigos_renipress)}")
print(f"CÓDIGOS EN ATENCIONES SIN MATCH EN RENIPRESS: {len(sin_match)}")
print(f"PORCENTAJE SIN MATCH: {len(sin_match)/len(codigos_atenciones)*100:.2f}%")
```
